### Text is a collection of meaningful sentences. Each sentence in turn comprises many words.

```python
import nltk
from nltk import *
sent = "India is a republic nation. We are proud Indians"
```

Texts are represented in Python in the form lists:  ['India is a republic nation. We are proud Indians']

 We can use slicing, indexing and we can also find length of these lists.

```python
print(len(sent))   #Prints the number of characters
print(sent[0:5])   #Prints 'India'
print(sent[11:19]) #Prints 'republic'
```

## Tokens

A meaningful unit of text is a token. Words are usually considered as tokens in NLP. The process of breaking a text based on the token is called tokenization.
